```python

import os
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
import numpy as np
from scipy.stats import skew, kurtosis
import mne
import matplotlib.pyplot as plt

# ==================== CONFIGURATION ====================
data_dir = r"/BCICIV_2b_gdf"
train_files = ['B0101T.gdf', 'B0102T.gdf','B0201T.gdf', 'B0202T.gdf','B0301T.gdf', 'B0302T.gdf','B0401T.gdf', 'B0402T.gdf','B0501T.gdf', 'B0502T.gdf','B0601T.gdf', 'B0602T.gdf','B0701T.gdf', 'B0702T.gdf','B0801T.gdf', 'B0802T.gdf','B0901T.gdf', 'B0902T.gdf']
eval_file = 'B0103T.gdf'

# Filter parameters
lpf_cutoff = 30.0
hpf_cutoff = 8.0
filter_order = 2
mav_window_ms = 100

# Epoch parameters
baseline_duration = 2.0
test_start = 0.5
test_duration = 3.0

# ==================== DEFINE PROCESSING FUNCTIONS ====================
def load_and_process_file(filepath, hpf_cutoff, lpf_cutoff, filter_order, mav_window_ms):
    """Load and process a single EEG file"""
    print(f"\nLoading: {filepath}")
    raw = mne.io.read_raw_gdf(filepath, preload=True)
    
    # Drop EOG channels
    eog_channels = ['EOG:ch01', 'EOG:ch02', 'EOG:ch03']
    raw.drop_channels([ch for ch in eog_channels if ch in raw.ch_names], on_missing='ignore')
    
    Fs = raw.info['sfreq']
    
    # Get events
    events = mne.events_from_annotations(raw)
    events_array = events[0]
    event_dict = events[1]
    
    # Find left/right IDs
    left_id = None
    right_id = None
    for event_name, event_id in event_dict.items():
        name_lower = event_name.lower()
        if 'left' in name_lower or '769' in name_lower or event_id == 769: left_id = event_id
        elif 'right' in name_lower or '770' in name_lower or event_id == 770: right_id = event_id
    
    if left_id is None or right_id is None:
        for event_name, event_id in event_dict.items():
            if '1' in event_name: left_id = event_id
            elif '2' in event_name: right_id = event_id
    
    events_motor = events_array[(events_array[:, 2] == left_id) | (events_array[:, 2] == right_id)]
    
    # Filter butterworth
    def butterworth_lpf(signal, freq_cutoff, orde):
        omega_c = 2 * np.pi * freq_cutoff
        Ts = 1 / Fs
        Ts2 = Ts ** 2
        wc2 = omega_c ** 2
        y = np.zeros(len(signal))
        if orde == 2:
            k1 = 4/Ts2
            k2 = 2*np.sqrt(2)*omega_c/Ts
            norm = k1 + k2 + wc2
            b0 = wc2 / norm
            b1 = 2 * b0
            b2 = b0
            a1 = (2*wc2 - 2*k1) / norm
            a2 = (k1 - k2 + wc2) / norm
            for n in range(len(signal)):
                if n < 2: y[n] = b0*signal[n]
                else: y[n] = b0*signal[n] + b1*signal[n-1] + b2*signal[n-2] - a1*y[n-1] - a2*y[n-2]
        return y

    def butterworth_hpf(signal, freq_cutoff, orde):
        omega_c = 2 * np.pi * freq_cutoff
        Ts = 1 / Fs
        Ts2 = Ts**2
        wc2 = omega_c**2
        y = np.zeros(len(signal))
        if orde == 2:
            k1 = 4/Ts2
            k2 = 2*np.sqrt(2)*omega_c/Ts
            norm = wc2 + k2 + k1
            b0 = k1 / norm
            b1 = -2 * b0
            b2 = b0
            a1 = (2*wc2 - 2*k1) / norm
            a2 = (wc2 - k2 + k1) / norm
            for n in range(len(signal)):
                if n < 2: y[n] = b0*signal[n]
                else: y[n] = b0*signal[n] + b1*signal[n-1] + b2*signal[n-2] - a1*y[n-1] - a2*y[n-2]
        return y
    
    # Load and filter data
    data_raw, times = raw[:, :]
    data_raw = data_raw * 1e6
    
    channels_to_process = [0, 1, 2]  # C3, Cz, C4
    data_filtered = np.zeros((len(channels_to_process), data_raw.shape[1]))
    
    for idx, ch_idx in enumerate(channels_to_process):
        signal = data_raw[ch_idx, :]
        hpf_fwd = butterworth_hpf(signal, hpf_cutoff, filter_order)
        lpf_fwd = butterworth_lpf(hpf_fwd, lpf_cutoff, filter_order)
        hpf_bwd = butterworth_hpf(lpf_fwd[::-1], hpf_cutoff, filter_order)
        lpf_bwd = butterworth_lpf(hpf_bwd, lpf_cutoff, filter_order)
        data_filtered[idx, :] = lpf_bwd[::-1]
    
    # Power and smooth
    data_power = data_filtered ** 2
    
    def moving_average(signal, window_size):
        kernel = np.ones(window_size) / window_size
        smoothed = np.convolve(signal, kernel, mode='same')
        return smoothed
    
    mav_window_samples = int(mav_window_ms * Fs / 1000)
    data_smoothed = np.zeros_like(data_power)
    
    for ch_idx in range(data_power.shape[0]):
        data_smoothed[ch_idx, :] = moving_average(data_power[ch_idx, :], mav_window_samples)
    
    # Extract epochs
    baseline_samples = int(baseline_duration * Fs)
    test_start_samples = int(test_start * Fs)
    test_end_samples = int((test_start + test_duration) * Fs)
    
    events_left = events_motor[events_motor[:, 2] == left_id]
    events_right = events_motor[events_motor[:, 2] == right_id]
    
    epochs_left_test = []
    epochs_right_test = []
    
    for event in events_left:
        cue_sample = event[0]
        baseline_start = cue_sample - baseline_samples
        test_start_idx = cue_sample + test_start_samples
        test_end_idx = cue_sample + test_end_samples
        
        if baseline_start >= 0 and test_end_idx < data_smoothed.shape[1]:
            test_epoch = data_smoothed[:, test_start_idx:test_end_idx]
            epochs_left_test.append(test_epoch)
    
    for event in events_right:
        cue_sample = event[0]
        baseline_start = cue_sample - baseline_samples
        test_start_idx = cue_sample + test_start_samples
        test_end_idx = cue_sample + test_end_samples
        
        if baseline_start >= 0 and test_end_idx < data_smoothed.shape[1]:
            test_epoch = data_smoothed[:, test_start_idx:test_end_idx]
            epochs_right_test.append(test_epoch)
    
    epochs_left_test = np.array(epochs_left_test)
    epochs_right_test = np.array(epochs_right_test)
    
    return epochs_left_test, epochs_right_test, Fs

def extract_power_features(epochs_data):
    """Extract power + statistical features (mean, std, var, skewness, kurtosis)"""
    n_trials = epochs_data.shape[0]
    n_channels = epochs_data.shape[1]
    features = []
    
    for trial in range(n_trials):
        trial_features = []
        for ch in range(n_channels):
            signal = epochs_data[trial, ch, :]
            
            trial_features.extend([
                np.mean(signal),
                np.std(signal),
                np.var(signal),
                np.max(signal),
                np.min(signal),
                skew(signal),                 # ➕ skewness
                kurtosis(signal, fisher=True) # ➕ kurtosis
            ])
        features.append(trial_features)
    
    return np.array(features)

def extract_csp_and_power_features(epochs_left, epochs_right):
    """Extract CSP and power features"""
    from mne.decoding import CSP
    
    # Combine for CSP
    X_csp = np.vstack([epochs_left, epochs_right])
    y_csp = np.hstack([np.zeros(len(epochs_left)), np.ones(len(epochs_right))])
    
    # Apply CSP
    csp = CSP(n_components=4, log=True, reg=1e-8)
    csp_features = csp.fit_transform(X_csp, y_csp)
    
    power_features = extract_power_features(X_csp)
    
    # Combine
    combined_features = np.hstack([csp_features, power_features])
    labels = y_csp
    
    return combined_features, labels, csp


# Train: B0101T + B0102T
train_epochs_left_all = []
train_epochs_right_all = []

for train_file in train_files:
    filepath = os.path.join(data_dir, train_file)
    epochs_left, epochs_right, Fs = load_and_process_file(filepath, hpf_cutoff, lpf_cutoff, filter_order, mav_window_ms)
    train_epochs_left_all.append(epochs_left)
    train_epochs_right_all.append(epochs_right)
    print(f"  {train_file}: Left={len(epochs_left)}, Right={len(epochs_right)}")

train_epochs_left = np.vstack(train_epochs_left_all)
train_epochs_right = np.vstack(train_epochs_right_all)

# Eval: B0103T
eval_filepath = os.path.join(data_dir, eval_file)
eval_epochs_left, eval_epochs_right, Fs = load_and_process_file(eval_filepath, hpf_cutoff, lpf_cutoff, filter_order, mav_window_ms)

print(f"\n{eval_file}: Left={len(eval_epochs_left)}, Right={len(eval_epochs_right)}")

# ==================== EXTRACT FEATURES ====================
print("\n" + "="*70)
print("FEATURE EXTRACTION")
print("="*70)

# Train features
train_features, train_labels, csp = extract_csp_and_power_features(train_epochs_left, train_epochs_right)
# Eval features (using CSP fitted on train)
X_eval = np.vstack([eval_epochs_left, eval_epochs_right])
y_eval = np.hstack([np.zeros(len(eval_epochs_left)), np.ones(len(eval_epochs_right))])

csp_eval = csp.transform(X_eval)

power_eval = extract_power_features(X_eval)
eval_features = np.hstack([csp_eval, power_eval])
eval_features = np.hstack([csp_eval, power_eval])

# Standardize
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(train_features)
X_eval_scaled = scaler.transform(eval_features)

print(f"Train features: {X_train_scaled.shape}")
print(f"  Class 0 (Left): {np.sum(train_labels == 0)}")
print(f"  Class 1 (Right): {np.sum(train_labels == 1)}")
print(f"\nEval features: {X_eval_scaled.shape}")
print(f"  Class 0 (Left): {np.sum(y_eval == 0)}")
print(f"  Class 1 (Right): {np.sum(y_eval == 1)}")


rf_model = RandomForestClassifier(
    n_estimators=200,
    max_depth=15,
    min_samples_split=5,
    min_samples_leaf=2,
    random_state=42,
    n_jobs=-1,
    class_weight='balanced'
)

rf_model.fit(X_train_scaled, train_labels)

y_pred_rf = rf_model.predict(X_eval_scaled)
y_pred_proba_rf = rf_model.predict_proba(X_eval_scaled)

accuracy_rf = accuracy_score(y_eval, y_pred_rf)
conf_matrix_rf = confusion_matrix(y_eval, y_pred_rf)

print(f"\nEvaluation Accuracy: {accuracy_rf:.4f} ({accuracy_rf*100:.2f}%)")
print(f"Confusion Matrix:\n{conf_matrix_rf}")
print(f"\nClassification Report:")
print(classification_report(y_eval, y_pred_rf, target_names=['Left Hand', 'Right Hand']))

sensitivity_rf = conf_matrix_rf[1,1]/(conf_matrix_rf[1,1]+conf_matrix_rf[0,1])
specificity_rf = conf_matrix_rf[0,0]/(conf_matrix_rf[0,0]+conf_matrix_rf[1,0])

print(f"Sensitivity: {sensitivity_rf:.4f}")
print(f"Specificity: {specificity_rf:.4f}")

# ==================== VISUALIZATION ====================
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

# 1. Confusion Matrix
im = axes[0, 0].imshow(conf_matrix_rf, cmap='Blues', aspect='auto')
axes[0, 0].set_xticks([0, 1])
axes[0, 0].set_yticks([0, 1])
axes[0, 0].set_xticklabels(['Left', 'Right'])
axes[0, 0].set_yticklabels(['Left', 'Right'])
axes[0, 0].set_xlabel('Predicted', fontweight='bold')
axes[0, 0].set_ylabel('Actual', fontweight='bold')
axes[0, 0].set_title('RF Confusion Matrix (B0101T+B0102T → B0103T)', fontweight='bold', fontsize=12)
for i in range(2):
    for j in range(2):
        axes[0, 0].text(j, i, str(conf_matrix_rf[i, j]), 
                       ha='center', va='center', color='black', fontsize=14, fontweight='bold')
plt.colorbar(im, ax=axes[0, 0])

# 2. Confidence Distribution
left_confs_rf = y_pred_proba_rf[y_eval == 0, 0] * 100
right_confs_rf = y_pred_proba_rf[y_eval == 1, 1] * 100

axes[0, 1].hist(left_confs_rf, bins=15, alpha=0.6, label='Left Hand', color='red', edgecolor='black')
axes[0, 1].hist(right_confs_rf, bins=15, alpha=0.6, label='Right Hand', color='green', edgecolor='black')
axes[0, 1].set_xlabel('Confidence (%)', fontweight='bold')
axes[0, 1].set_ylabel('Frequency', fontweight='bold')
axes[0, 1].set_title('RF Prediction Confidence Distribution', fontweight='bold', fontsize=12)
axes[0, 1].legend(fontsize=10)
axes[0, 1].grid(True, alpha=0.3)

# 3. Per-Sample Confidence
axes[1, 0].scatter(np.where(y_eval == 0)[0], y_pred_proba_rf[y_eval == 0, 0], 
                   alpha=0.6, s=50, label='Left Hand', color='red')
axes[1, 0].scatter(np.where(y_eval == 1)[0], y_pred_proba_rf[y_eval == 1, 1], 
                   alpha=0.6, s=50, label='Right Hand', color='green')
axes[1, 0].axhline(y=0.5, color='black', linestyle='--', alpha=0.5, linewidth=2)
axes[1, 0].set_xlabel('Sample Index', fontweight='bold')
axes[1, 0].set_ylabel('Prediction Probability', fontweight='bold')
axes[1, 0].set_title('RF Per-Sample Prediction Confidence', fontweight='bold', fontsize=12)
axes[1, 0].legend(fontsize=10)
axes[1, 0].grid(True, alpha=0.3)
axes[1, 0].set_ylim([0, 1])

# 4. Performance Metrics
metrics_names = ['Accuracy', 'Sensitivity', 'Specificity']
metrics_values = [accuracy_rf, sensitivity_rf, specificity_rf]
colors_metrics = ['#1f77b4', '#ff7f0e', '#2ca02c']

axes[1, 1].bar(metrics_names, metrics_values, color=colors_metrics, alpha=0.7, edgecolor='black', linewidth=2)
axes[1, 1].set_ylabel('Score', fontweight='bold')
axes[1, 1].set_title('RF Performance Metrics', fontweight='bold', fontsize=12)
axes[1, 1].set_ylim([0, 1])
for i, (name, val) in enumerate(zip(metrics_names, metrics_values)):
    axes[1, 1].text(i, val + 0.03, f'{val*100:.2f}%', ha='center', fontsize=11, fontweight='bold')
axes[1, 1].grid(True, alpha=0.3, axis='y')

plt.tight_layout()
plt.show()

# ==================== SUMMARY ====================
print("\n" + "="*70)
print("RANDOM FOREST SUMMARY (B0101T+B0102T → B0103T)")
print("="*70)
print(f"Training Data: B0101T.gdf + B0102T.gdf")
print(f"Evaluation Data: B0103T.gdf")
print(f"Total Features: {X_train_scaled.shape[1]}")
print(f"Training Samples: {X_train_scaled.shape[0]}")
print(f"Evaluation Samples: {X_eval_scaled.shape[0]}")
print(f"Evaluation Accuracy: {accuracy_rf*100:.2f}%")
print(f"Mean Confidence (Left): {np.mean(left_confs_rf):.2f}%")
print(f"Mean Confidence (Right): {np.mean(right_confs_rf):.2f}%")
print(f"Sensitivity: {sensitivity_rf:.4f}")
print(f"Specificity: {specificity_rf:.4f}")
print(f"Train features: {X_train_scaled.shape}")  # (186, 22)
print(f"  Class 0 (Left): {np.sum(train_labels == 0)}")   # 93
print(f"  Class 1 (Right): {np.sum(train_labels == 1)}")  # 93
print("="*70)
```
